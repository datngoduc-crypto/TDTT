Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
a=float(input())
b=float(input())
c=float(input())
if a>0 and b>0 and c>0 and a+b>c and b+c>a and a+c>b:
    if a==b==c:
        print("Tam giác đều")
    elif a==b or b==c or a==c:
        print("Tam giác cân")
    elif a**2==b**2+c**2 or b**2==a**2+c**2 or c**2==a**2+b**2:
        print("Tam giác vuông")
    else:
        print("Tam giác thường")
else:
    print("Khong la ba canh ca mot tam giac")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
import math
a=float(input())
b=float(input())
c=float(input())
if a==0:
    if b==0:
        if c==0:
            print("Phương trình vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        x=-c/b
        print(f"Phương trình có nghiệm là:{x}")
else:
    delta=b**2-4*a*c
    if delta <0:
        print("Phương trình vô nghiệm")
    elif delta==0:
        x=-b/(2*a)
        print("Phương trình có nghiệm kép: x1=x2={x}")
    else:
        x1=(-b+math.sqrt(delta))/(2*a)
        x2=(-b-math.sqrt(delta))/(2*a)
        print(f"Phương trình có hai nghiệm phân biệt: x1={x1},x2={x2}")
```


Bài 8: Giải hệ phương trình tuyến tính:
ax+by=m và cx+dy=n

```python
a=float(input("Nhập a:"))
b=float(input("Nhập b:"))
m=float(input("Nhập m:"))
c=float(input("Nhập c:"))
d=float(input("Nhập d:"))
n=float(input("Nhập n:"))
D=a*d-b*c
Dx=m*d-b*n
Dy=a*n-m*c
if D!=0:
    x=Dx/D
    y=Dy/D
    print(f"Hệ phương trình có nghiệm duy nhất:x={x}, y={y}")
else:
    if Dx!=0 or Dy!=0:
        print("Hệ phương trình vô nghiệm")
    else:
        print("Hệ phương trình vô số nghiệm")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
import math
x0,y0=map(float,input("Nhập tọa độ tâm I").split())
R=float(input("Nhập bán kính R: "))
x,y=map(float,input("Nhập tọa độ điểm A").split())
d=math.sqrt((x-x0)**2+(y-y0)**2)
if d==R:
    print("Điểm A nằm trên đường tròn")
elif d>R:
    print("A nằm ngoài đường tròn")
elif d<R:
    print("A nằm trong đường tròn")
```
