W3A1. Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý

```python
n=int(input("Nhập số nguyên dương: "))
if 100<=n<=999:
    tram=n//100
    chuc=(n//10)%10
    don_vi=n%10
    so_dao_nguoc=don_vi*100+chuc*10+tram
    print(f"Số đảo ngược:{so_dao_nguoc}")
else:
    print("Vui lòng nhập lại")
```

W3A3 Kiểm tra xem một số có phải là lũy thừa của 2 hay không. Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không.
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0

```python
n=int(input())
if n>=0 and n&(n-1)==0:
    print("Đây là lũy thừa của 2")
else:
    print("Đây không là lũy thừa của 2")
```
